```python
#Exercicio 3

# Calculo de aumento de salário

Salario = float(input("Digite o valor do seu salario incluindo casas decimais"))
PercentualAumento = float(input("Qual o valor percentual de aumento? "))

PercentualDecimal = PercentualAumento / 100

SalarioCalculado = Salario + (Salario * PercentualDecimal)

print(f"O valor do seu salario atualizado é de : {SalarioCalculado}")
```

```python
#Exercicio 4

# Calculo de volume de uma lata

 PI = 3.14

Raio = float(input("Digite o valor do raio lata em centimetros"))
Altura = float(input("Digite o valor da altura da lata em centimetros"))

Volume = PI * (Raio**2) * Altura

print(f"O valor do volume da lata é de : {Volume} cm")
```
